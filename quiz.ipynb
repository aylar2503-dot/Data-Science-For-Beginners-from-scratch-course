### Вопрос 3
Зачем нужна кнопка History?

Ответ:
History показывает историю коммитов — сохранённых изменений проекта.

Если нажать правой кнопкой мыши на коммит, можно:
- Create branch from commit — создать ветку.
- Revert changes — отменить изменения коммита.
- Reset to commit — вернуть ветку к выбранному коммиту.
- Checkout commit — посмотреть состояние проекта на момент коммита.
- Copy SHA — скопировать идентификатор коммита.
- View on GitHub — открыть коммит на GitHub.
